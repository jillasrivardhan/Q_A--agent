```python
from langchain_groq import ChatGroq
from langchain_community.utilities import GoogleSerperAPIWrapper
from langgraph.checkpoint.memory import MemorySaver 
from langchain.agents import create_agent 
```

```python
model = ChatGroq(
   model = "openai/gpt-oss-120b",
)
```

```python
search = GoogleSerperAPIWrapper()
tool = search.run   
```

```python
agent = create_agent(
   model = model,
   tools = [tool],
   checkpointer = MemorySaver(),
   system_prompt = "You are a helpful assistant that can answer questions based on the information provided by the user. If you don't know the answer, just say you don't know. Do not try to make up an answer.you can also search the web.",
)
```

```python
question = input("user: ")
response = agent.invoke({"messages":[{"role":"user","content":question}]},
             {"configurable": {"thread_id": "1"},
              })
```
